## Dummy firms

Creates `pg.bas_firms.firm_add_dummy`: same schema as `firm_add_week_37` plus a
`is_prospect boolean` column, filled with 20 generated rows (`padoa_id` 999000001-999000020).
`create or replace` makes the cell re-runnable.

```python
DUMMY_TABLE = "pg.bas_firms.firm_add_dummy"

duck.sql(f"""
create or replace table {DUMMY_TABLE} as
with l as (
    select
        ['Nordlicht Logistik GmbH', 'Bergmann Metallbau GmbH & Co. KG', 'Sonnenhof Pflegedienst gGmbH',
         'Weber Elektrotechnik GmbH', 'Rheinfels Bau AG', 'Kranich Gastronomie GmbH',
         'Talblick Gartenbau GmbH', 'Hansa Reinigungsservice GmbH', 'Eichberg Kunststoffe GmbH',
         'Morgenstern Kita gGmbH', 'Silberbach Spedition GmbH', 'Adlerhorst Sicherheitsdienst GmbH',
         'Lindenhof Seniorenzentrum gGmbH', 'Falkenstein Maschinenbau GmbH', 'Uferwerk Druckerei GmbH',
         'Kastanie Bäckerei GmbH', 'Schwanensee Hotelbetriebs GmbH', 'Tannenweg IT Services GmbH',
         'Rosental Textilhandel GmbH', 'Möwenflug Fischhandel GmbH'] as names,
        ['Zentrale', 'Werk Nord', 'Filiale Mitte', 'Standort Süd', 'Lager West'] as sites,
        ['Berlin', 'Hamburg', 'Köln', 'München', 'Leipzig'] as cities,
        ['10115', '20095', '50667', '80331', '04109'] as post_codes,
        ['Hauptstraße', 'Lindenallee', 'Bahnhofstraße', 'Gartenweg', 'Industriering'] as streets,
        ['Berlin', 'Hamburg', 'Köln'] as standorte,
        ['direct', 'indirect'] as directions,
        ['to create', 'to update', 'done'] as statuses,
        ['Team Nord', 'Team Süd', 'Team West'] as med_teams,
        ['Anna Schmidt', 'Ben Krüger', 'Clara Vogt'] as owners,
        ['Dirk Hoffmann', 'Eva Lange', 'Felix Braun'] as experts
),
i as (select unnest(generate_series(1, 20))::integer as n)
select
    'dummy row ' || i.n                                          as "comment",
    'check dummy data'                                           as "To do",
    (999000000 + i.n)::bigint                                    as padoa_id,
    l.names[i.n]                                                 as "name",
    l.sites[(i.n - 1) % len(l.sites) + 1]                        as "commercial name",
    strftime(date '2026-01-06' + (i.n * 7), '%Y-%m-%d')           as enrolment_date,
    l.standorte[(i.n - 1) % len(l.standorte) + 1]                as "BAS standort",
    l.streets[(i.n - 1) % len(l.streets) + 1] || ' ' || (i.n + 1) as a_street,
    l.post_codes[(i.n - 1) % len(l.post_codes) + 1]              as a_post_code,
    l.cities[(i.n - 1) % len(l.cities) + 1]                      as a_city,
    l.streets[i.n % len(l.streets) + 1] || ' ' || (i.n + 20)     as h_street,
    l.post_codes[i.n % len(l.post_codes) + 1]                    as h_post_code,
    l.cities[i.n % len(l.cities) + 1]                            as h_city,
    null::varchar                                                as b_street,
    null::varchar                                                as b_post_code,
    null::varchar                                                as b_city,
    lpad(((i.n * 37) % 99 + 1)::varchar, 2, '0') || '.1'         as "WZ",
    'BG' || ((i.n % 9) + 1)::varchar                             as "BG",
    ((i.n * 13) % 240 + 10)::varchar                             as "no. employees in basic care",
    ((i.n * 3) % 80 + 5)::varchar                                as "basic care hrs total",
    ((i.n * 2) % 40 + 2)::varchar                                as "basic care hrs arbmed",
    ((i.n * 2) % 30 + 3)::varchar                                as "basic care hrs arbsich",
    l.directions[(i.n - 1) % len(l.directions) + 1]              as "direct / indirect",
    l.med_teams[(i.n - 1) % len(l.med_teams) + 1]                as "medical team",
    l.owners[(i.n - 1) % len(l.owners) + 1]                      as "account owner",
    l.experts[(i.n - 1) % len(l.experts) + 1]                    as "safety expert",
    l.statuses[(i.n - 1) % len(l.statuses) + 1]                  as "status",
    case when i.n % 5 = 0 then 'DUMMY-MOTHER-' || i.n end        as "acc.coll mother",
    case when i.n % 5 = 1 then 'DUMMY-MOTHER-' || (i.n - 1) end  as "acc.coll daughter",
    case when i.n % 7 = 0 then 'yes' end                         as "simple coll",
    case when i.n % 4 = 0 then 'yes' end                         as "admin shell (basic care invoicing)",
    null::varchar                                                as "merge into",
    null::varchar                                                as comment_1,
    (i.n % 3 = 0)                                                as is_prospect
from i, l
order by i.n
""")
```

```python
duck.sql(f"""
    select count(*) as rows, count(*) filter (where is_prospect) as prospects
    from {DUMMY_TABLE}
""")
```

```python
duck.sql(f'select * from {DUMMY_TABLE} order by padoa_id')
```
